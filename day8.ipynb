```python
# Line Chart
# Line Chart
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Name": ["Reshma", "Shiva", "Sowjanya", "Rishi", "Jhansi"],
    "Age": [18, 17, 18, 19, 20],
    "Marks": [85, 96, 85, 90, 88]
}
df = pd.DataFrame(data)
sns.lineplot(x=df.index, y="Age", data=df)
plt.xlabel("Index")
plt.ylabel("Age")
plt.title("Age Line Plot")
plt.show()
```

```python
#Scatter plot
sns.scatterplot(x="Age", y="Marks", data=df)

plt.xlabel("Age")
plt.ylabel("Marks")
plt.title("Age vs Marks")
plt.show()
```

```python
#Box plot
sns.boxplot(y="Marks", data=df)

plt.title("Box Plot of Marks")
plt.show()
```

```python
#Violin Plot
sns.violinplot(y="Marks", data=df)

plt.title("Violin Plot of Marks")
plt.show()
```

```python
#Swarm Plot
sns.swarmplot(x="Name", y="Marks", data=df)

plt.title("Swarm Plot")
plt.show()
```

```python
#Bar Plot

sns.barplot(x="Name", y="Marks", data=df)

plt.title("Bar Plot")
plt.show()
```

```python
#Point Plot
sns.pointplot(x="Name", y="Marks", data=df)

plt.title("Point Plot")
plt.show()
```

```python
#Count Plot
sns.countplot(x="Name", data=df)

plt.title("Count Plot")
plt.show()
```

```python
#KDE(Kernal Density Estimate)
sns.kdeplot(x="Marks", data=df, fill=True)

plt.title("KDE Plot")
plt.show()
```

```python
#Adding Titles and Axis Labels

sns.scatterplot(x="Age", y="Marks", data=df)

plt.title("Age vs Marks")
plt.xlabel("Age")
plt.ylabel("Marks")

plt.show()
```

```python
#Built-in Styles and Grids in Seaborn
sns.set_style("whitegrid")

sns.boxplot(x="Name", y="Marks", data=df)
data=sns.load_dataset("iris")
plt.title("Marks Distribution")
plt.show()
```

```python
#Customizing Color Palettes
iris = sns.load_dataset("iris")

sns.set_palette("pastel")

sns.violinplot(x="species", y="petal_length", data=iris)

plt.title("Petal Length by Species")
plt.show()
```

```python
#Using a custom palette
import seaborn as sns
import matplotlib.pyplot as plt

iris = sns.load_dataset("iris")

custom_colors = ["crimson", "gold", "deepskyblue"]

sns.set_palette(custom_colors)

sns.violinplot(x="species", y="petal_length", data=iris)

plt.title("Custom Colored Petal Length ")
plt.show()
```

```python
#Adjusting Figure Size and Aspect Ratio
plt.figure(figsize=(10, 5))
sns.lineplot(x='Name', y='Marks', data=df, marker='o')

plt.title('Student Marks')
plt.xlabel('Name')
plt.ylabel('Marks')
plt.show()
```

```python
#Adding Markers to Line Plot
plt.figure(figsize=(8, 5))
plt.plot(df["Name"], df["Marks"], marker='o')

plt.title("Student Marks")
plt.xlabel("Name")
plt.ylabel("Marks")
plt.show()
```

```python
#Pair Plots
iris = sns.load_dataset("iris")

# Pair Plot
sns.pairplot(iris, hue="species")

plt.show()
```

```python
#Joint Plots
iris = sns.load_dataset("iris")
sns.jointplot(
    data=iris,
    x="sepal_length",y="sepal_width",kind="scatter"
)
plt.show()
```

```python
#Grid Plot
iris = sns.load_dataset("iris")
g = sns.PairGrid(iris, hue="species")
g.map_diag(sns.histplot)
g.map_offdiag(sns.scatterplot)
g.add_legend()
plt.show()
```

```python
#Regression Plots: Visualizing Linear Relationships
iris = sns.load_dataset("iris")
sns.lmplot(
    data=iris,x="sepal_length",y="petal_length",hue="species"
)

plt.show()
```
