```python
#step 1: Import libraries
import pandas as pd
import numpy as np
```

```python
# Step 2: Load dataset
data = {
    'Name': ['Reshma', 'Ayesha', 'Reshma', None],
    'Age': [25, np.nan, 25, 100],
    'Score': [80, 90, 80, 100]
}
df = pd.DataFrame(data)

print("Original Data:")
print(df)
```

```python
# Step 3: Remove duplicates
df = df.drop_duplicates()
print(df)
```

```python
# Step 4: Handle missing values
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Name'] = df['Name'].fillna('Unknown')
print(df)
```

```python
# Step 5: Detect outliers (using mean ± 2*std)
mean = df['Age'].mean()
std = df['Age'].std()

lower = mean - 2 * std
upper = mean + 2 * std
print("Lower value:", lower, "Upper Value:", upper)
df = df[(df['Age'] >= lower) & (df['Age'] <= upper)]
print(df)
```

```python
# Step 6: Display cleaned data
print("\nCleaned Data:")
print(df)
```
