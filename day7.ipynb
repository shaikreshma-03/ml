```python
# Import Libraries
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
# Create DataFrame
df = pd.DataFrame({
    "Name": ["Reshma", "Sony", "Reshma", "Ayesha"],
    "Age": [18, 19, 18, 19],
    "Marks": [85, 96, 85, 81]
})

print(df)
```

```python
#line chart
plt.plot(df["Name"], df["Marks"], marker='o')

plt.title("Name vs Marks")
plt.xlabel("Name")
plt.ylabel("Marks")

plt.show()
```

```python
#Bar Chart
plt.bar(df["Name"], df["Marks"], color='skyblue')

plt.title("Students Marks")
plt.xlabel("Name")
plt.ylabel("Marks")

plt.show()
```

```python
#Histogram
plt.hist(df["Marks"], bins=5, color='orange', edgecolor='black')

plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")

plt.show()
```

```python
#Scatter graph
plt.scatter(df["Age"], df["Marks"], color='red')

plt.title("Age vs Marks")
plt.xlabel("Age")
plt.ylabel("Marks")

plt.show()
```

```python
#Pie chart
plt.pie(df["Marks"],
        labels=df["Name"],
        autopct='%1.1f%%')

plt.title("Marks Percentage")

plt.show()
```

```python
#Bar plot
plt.boxplot(df["Marks"])

plt.title("Box Plot of Marks")
plt.ylabel("Marks")

plt.show()
```

```python
#Heatmap
data = df[["Age", "Marks"]].corr()

plt.imshow(data, cmap='viridis')

plt.colorbar()

plt.xticks([0,1], ["Age", "Marks"])
plt.yticks([0,1], ["Age", "Marks"])

plt.title("Heatmap")

plt.show()
```

```python
#Customized Line chart
x = [10, 20, 30, 40]
y = [20, 25, 35, 55]

plt.plot(x, y,
         color='green',
         linewidth=3,
         marker='o',
         markersize=10,
         linestyle='--')

plt.title("Customized Line Chart")
plt.xlabel("X-Axis")
plt.ylabel("Y-Axis")
plt.show()
```

```python
#Customized Bar chart
x = ['Thur', 'Fri', 'Sat', 'Sun']
y = [170, 120, 250, 190]

plt.bar(x, y,
        color='green',
        edgecolor='black',
        linewidth=2)

plt.title("Customized Bar Chart")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.show()
```

```python
#Customized Histogram
x = [7, 8, 9, 10, 10, 11, 12, 12, 13, 14, 14, 15,
     16, 16, 17, 18, 18, 19, 20, 20, 21, 22, 23,
     24, 25, 25, 26, 28, 30, 32, 35, 36, 38, 40,
     42, 44, 48, 50]
plt.hist(x, bins=10,color='green',edgecolor='blue',linestyle='--',alpha=0.5)
plt.title("Customized Histogram")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

```python
#Customized Scatter Plot

# Create DataFrame
df = pd.DataFrame({
    "Name": ["Reshma", "Sowji", "Shiva", "Rishi"],
    "Age": [18, 19, 20, 21],
    "Marks": [85, 90, 95, 88]
})
plt.scatter(df["Age"],df["Marks"],s=[150, 250, 300, 200],c=['red', 'blue', 'green', 'orange'],marker='D',edgecolor='black',linewidth=2,alpha=0.8)
plt.title("Age vs Marks")
plt.xlabel("Age")
plt.ylabel("Marks")

plt.show()
```

```python

# Customizing Pie Chart
explode = [0.1, 0, 0.2, 0]

colors = ["pink", "skyblue", "lightgreen", "orange"]

plt.pie(df["Marks"],
        labels=df["Name"],
        explode=explode,
        autopct='%1.1f%%',
        colors=colors,
        shadow=True)

plt.title("Customized Pie Chart")

plt.show()
```

```python
#Axes Class
x = df["Name"]
y = df["Marks"]

fig = plt.figure(figsize=(6,4))

ax = fig.add_axes([0.1, 0.1, 0.8, 0.8])

ax.plot(x, y,color='green',marker='o',linewidth=3,linestyle='--')
ax.set_title("Axes Graph")
ax.set_xlabel("Student Names")
ax.set_ylabel("Marks")

ax.legend(["Marks"])

plt.show()
```

```python
#Using add_axes()
x = df["Name"]
y = df["Marks"]
fig = plt.figure(figsize=(10, 4))
ax1 = fig.add_axes([0.1, 0.1, 0.35, 0.8])
ax2 = fig.add_axes([0.55, 0.1, 0.35, 0.8])
ax1.plot(x, y)
ax2.plot(y, x)
plt.show()
```

```python
#Using subplot()
x = df["Name"]
y = df["Marks"]
plt.figure(figsize=(10,4))
plt.subplot(121)
plt.plot(x, y)
plt.title("X vs Y")
plt.subplot(122)
plt.plot(y, x)
plt.title("Y vs X")
plt.show()
```

```python
#Using subplot2grid()
x = df["Name"]
y = df["Marks"]
axes1 = plt.subplot2grid((7, 1), (0, 0),rowspan=2,colspan=1)
axes2 = plt.subplot2grid((7, 1), (3, 0),rowspan=2,colspan=1)
axes1.plot(x, y)
axes1.set_title("X vs Y")
axes2.plot(y, x)
axes2.set_title("Y vs X")
plt.show()
```

```python
#Saving Plots Using savefig()
x = df["Name"]
y = df["Marks"]
plt.bar(x, y, color='skyblue')
plt.title("Students Marks")
plt.xlabel("Names")
plt.ylabel("Marks")
plt.savefig("students_marks.png")
plt.show()
```
